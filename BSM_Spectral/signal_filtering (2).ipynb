```python
def get_sp100():
    url = "https://en.wikipedia.org/wiki/S%26P_100"
    weblink = requests.get(url, headers=headers)
    tables = pd.read_html(weblink.text)
    tickers = tables[2]["Symbol"].tolist()
    return tickers
    
tickers = get_sp100()
```

```python
tickers
```

```python
import yfinance as yf
def download_sp100_data(start_date, end_date):
        data = yf.download(
            tickers=tickers,
            start=start_date,
            end=end_date,
            group_by='ticker',
            auto_adjust=True,
            threads=True
        )
        return data
```

```python
data = download_sp100_data("2013-01-01","2019-01-01")
```

```python
import numpy as np
```

```python
T = 1510
N = 100 
```

```python
lambra_ra_calculated = T/N
```

```python
#marchenko-pastur filtering
volatility=1
lambda_ra=lambra_ra_calculated
def marchenkop_filter_boundaries(standard_error, lambda_ra):
    lambda_plus = (volatility**2)*((1+np.sqrt(lambda_ra))**2)
    lambda_minus = (volatility**2)*((1-np.sqrt(lambda_ra))**2)
    return lambda_plus, lambda_minus

def marchenkop_filter_distr(lambda_plus, lambda_minus, p):
    marchenkop_val = (1/2*np.pi*volatility**2)*(np.sqrt((lambda_plus-p)*(p-lambda_minus))/lambda_ra*p)
    return marchenkop_val
```

```python
lambda_plus, lambda_minus = marchenkop_filter_boundaries(standard_error=volatility, lambda_ra=lambra_ra_calculated)
print(lambda_plus)
print(lambda_minus)
```

```python
for val in range(0,10000):
    marchenkop_val=marchenkop_filter_distr(lambda_plus, lambda_minus, val)
    print(marchenkop_val)
```
